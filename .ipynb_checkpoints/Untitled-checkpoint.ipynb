```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import importlib
import rioxarray
import rasterio
import matplotlib.colors as mcolors
import matplotlib
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

data_dir = "/scratch/hydro4/users/kv25483/FDRI/Rain_gauge_optimisation/Data/"
catchment_Dolwen = gpd.read_file(data_dir + "54080/54080.shp")

import functions
importlib.reload(functions)
from functions import *

# Load rain gauge metadata
rain_meta_cmd = pd.read_csv(data_dir + "rain_meta_for_CMD.csv")
rain_meta_sf = gpd.GeoDataFrame(
    rain_meta_cmd,
    geometry=gpd.points_from_xy(rain_meta_cmd["x_2"], rain_meta_cmd["y_2"]),
    crs="EPSG:27700"
)

rainfall_gdf_merged = gpd.read_file(data_dir + "Derived/rainfall_gdf_merged.gpkg")

# --------------------------------------------------
# 1. Load DEM
# --------------------------------------------------
dem_path = data_dir + "dem_Dolwen.tif"
with rioxarray.open_rasterio(dem_path, masked=True) as dem_rx:
    dem = dem_rx.squeeze()
    dem_arr = dem.values.astype(float)
    transform = dem.rio.transform()

dx = transform.a
dy = -transform.e

# --------------------------------------------------
# 2. Slope + aspect
# --------------------------------------------------
dz_dy, dz_dx = np.gradient(dem_arr, dy, dx)
slope_arr = np.degrees(np.arctan(np.sqrt(dz_dx**2 + dz_dy**2)))
aspect_arr = np.degrees(np.arctan2(-dz_dx, dz_dy))
aspect_arr = np.where(aspect_arr < 0, 360 + aspect_arr, aspect_arr)

# --------------------------------------------------
# 3. Build valid pixel mask + feature matrix
# --------------------------------------------------
valid_mask = ~np.isnan(dem_arr) & ~np.isnan(slope_arr) & ~np.isnan(aspect_arr)

aspect_rad = np.deg2rad(aspect_arr)
aspect_sin = np.sin(aspect_rad)
aspect_cos = np.cos(aspect_rad)

features = np.column_stack([
    dem_arr[valid_mask],
    slope_arr[valid_mask],
#     aspect_sin[valid_mask],
#     aspect_cos[valid_mask],
])

# --------------------------------------------------
# 4. Scale + cluster
# --------------------------------------------------
scaler = StandardScaler()
X = scaler.fit_transform(features)

k = 6
kmeans = KMeans(n_clusters=k, random_state=0)
labels = kmeans.fit_predict(X)

# --------------------------------------------------
# 5. Reshape labels back into 2D grid
# --------------------------------------------------
label_grid = np.full(dem_arr.shape, np.nan)
label_grid[valid_mask] = labels

label_xr = dem.copy()
label_xr.data = label_grid

# --------------------------------------------------
# 6. Plot
# --------------------------------------------------
cmap = matplotlib.colormaps["Set1"].resampled(k)
bounds = np.arange(-0.5, k, 1)
norm = mcolors.BoundaryNorm(bounds, cmap.N)

fig, ax = plt.subplots(figsize=(6, 5))
label_xr.plot(ax=ax, cmap=cmap, norm=norm, add_colorbar=True,
              cbar_kwargs={"ticks": np.arange(k), "label": "cluster"})
catchment_Dolwen.boundary.plot(ax=ax, color="black")
ax.axis('off')
ax.set_title(f'K-means clusters (k={k}), native resolution')
```

```python
from scipy.spatial.distance import cdist
import random

# --------------------------------------------------
# 1. Build ranked candidate lists per cluster
#    (top-N pixels closest to centroid in scaled feature space)
# --------------------------------------------------
N_CANDIDATES = 200   # native res has many more pixels than 1km cells — allow a wide pool
                      # so the greedy search has spatial room to satisfy min_dist

valid_rows, valid_cols = np.where(valid_mask)
rep_xs_all, rep_ys_all = rasterio.transform.xy(transform, valid_rows, valid_cols)
rep_xs_all = np.array(rep_xs_all)
rep_ys_all = np.array(rep_ys_all)

cluster_candidates = {}
cell_coords = {}

for cluster_id in range(k):
    cluster_mask = labels == cluster_id
    cluster_points = X[cluster_mask]
    global_indices = np.where(cluster_mask)[0]

    dists = cdist(cluster_points, kmeans.cluster_centers_[cluster_id].reshape(1, -1)).flatten()
    order = np.argsort(dists)
    top_n_global = global_indices[order[:N_CANDIDATES]]

    cluster_candidates[cluster_id] = list(top_n_global)

    for idx in top_n_global:
        cell_coords[idx] = (rep_xs_all[idx], rep_ys_all[idx])

# --------------------------------------------------
# 2. Run greedy / shuffle-greedy selection
# --------------------------------------------------
min_dist = 2000   # metres — adjust to your required gauge spacing
existing_coords = []   # or your existing gauge coordinates as (x, y) tuples

solution = shuffle_greedy(cluster_candidates, cell_coords, existing_coords,
                           min_dist, n_iterations=500)

# check whether every cluster actually got a feasible site
_, failed = greedy_selection(cluster_candidates, cell_coords, existing_coords, min_dist)
if failed:
    print("Clusters with no feasible candidate:", failed)

# --------------------------------------------------
# 3. Map chosen indices back to row/col + attributes
# --------------------------------------------------
selected_records = []
for cluster_id, idx in solution.items():
    row, col = valid_rows[idx], valid_cols[idx]
    x, y = cell_coords[idx]
    selected_records.append({
        "cluster": cluster_id,
        "row": row, "col": col,
        "x": x, "y": y,
        "dem": dem_arr[row, col],
        "slope": slope_arr[row, col],
        "aspect": aspect_arr[row, col],
    })

selected_df = pd.DataFrame(selected_records)
selected_df

# --------------------------------------------------
# 4. Plot
# --------------------------------------------------
fig, ax = plt.subplots(figsize=(6, 5))
label_xr.plot(ax=ax, cmap=cmap, norm=norm, add_colorbar=True,
              cbar_kwargs={"ticks": np.arange(k), "label": "cluster"})
catchment_Dolwen.boundary.plot(ax=ax, color="black")
ax.scatter(selected_df["x"], selected_df["y"], c="black", edgecolor="white",
           s=80, marker="*", zorder=5)
ax.axis('off')
ax.set_title(f'K-means clusters (k={k}) — spacing-constrained representatives')
```

```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import importlib
import rioxarray
import rasterio
import matplotlib.colors as mcolors
import matplotlib
import random
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist

data_dir = "/scratch/hydro4/users/kv25483/FDRI/Rain_gauge_optimisation/Data/"
catchment_Dolwen = gpd.read_file(data_dir + "54080/54080.shp")

import functions
importlib.reload(functions)
from functions import *

# Load rain gauge metadata
rain_meta_cmd = pd.read_csv(data_dir + "rain_meta_for_CMD.csv")
rain_meta_sf = gpd.GeoDataFrame(
    rain_meta_cmd,
    geometry=gpd.points_from_xy(rain_meta_cmd["x_2"], rain_meta_cmd["y_2"]),
    crs="EPSG:27700"
)

rainfall_gdf_merged = gpd.read_file(data_dir + "Derived/rainfall_gdf_merged.gpkg")

# --------------------------------------------------
# 1. Load DEM
# --------------------------------------------------
dem_path = data_dir + "dem_Dolwen.tif"
with rioxarray.open_rasterio(dem_path, masked=True) as dem_rx:
    dem = dem_rx.squeeze()
    dem_arr = dem.values.astype(float)
    transform = dem.rio.transform()

dx = transform.a
dy = -transform.e

# --------------------------------------------------
# 2. Slope + aspect
# --------------------------------------------------
dz_dy, dz_dx = np.gradient(dem_arr, dy, dx)
slope_arr = np.degrees(np.arctan(np.sqrt(dz_dx**2 + dz_dy**2)))
aspect_arr = np.degrees(np.arctan2(-dz_dx, dz_dy))
aspect_arr = np.where(aspect_arr < 0, 360 + aspect_arr, aspect_arr)

# --------------------------------------------------
# 3. Build valid pixel mask + feature matrix
# --------------------------------------------------
valid_mask = ~np.isnan(dem_arr) & ~np.isnan(slope_arr) & ~np.isnan(aspect_arr)

aspect_rad = np.deg2rad(aspect_arr)
aspect_sin = np.sin(aspect_rad)
aspect_cos = np.cos(aspect_rad)

features = np.column_stack([
    dem_arr[valid_mask],
    slope_arr[valid_mask],
    aspect_sin[valid_mask],
    aspect_cos[valid_mask],
])

# --------------------------------------------------
# 4. Scale + cluster
# --------------------------------------------------
scaler = StandardScaler()
X = scaler.fit_transform(features)

k = 6
kmeans = KMeans(n_clusters=k, random_state=0)
labels = kmeans.fit_predict(X)

# --------------------------------------------------
# 5. Reshape labels back into 2D grid
# --------------------------------------------------
label_grid = np.full(dem_arr.shape, np.nan)
label_grid[valid_mask] = labels

label_xr = dem.copy()
label_xr.data = label_grid

# --------------------------------------------------
# 6. Plot clusters
# --------------------------------------------------
cmap = matplotlib.colormaps["Set1"].resampled(k)
bounds = np.arange(-0.5, k, 1)
norm = mcolors.BoundaryNorm(bounds, cmap.N)

fig, ax = plt.subplots(figsize=(6, 5))
label_xr.plot(ax=ax, cmap=cmap, norm=norm, add_colorbar=True,
              cbar_kwargs={"ticks": np.arange(k), "label": "cluster"})
catchment_Dolwen.boundary.plot(ax=ax, color="black")
ax.axis('off')
ax.set_title(f'K-means clusters (k={k}), native resolution')

# --------------------------------------------------
# 7. Build candidate lists per cluster + array-based cell_coords
#    (cell_coords MUST be an array — greedy_selection/is_feasible
#    use numpy fancy indexing: cell_coords[selected_idx])
# --------------------------------------------------
N_CANDIDATES = 200   # size of candidate pool per cluster; increase if clusters fail

valid_rows, valid_cols = np.where(valid_mask)
rep_xs_all, rep_ys_all = rasterio.transform.xy(transform, valid_rows, valid_cols)

# (n_valid_pixels, 2) array — row i corresponds to valid_rows[i], valid_cols[i],
# and to the same "global index" used in X / labels
cell_coords = np.column_stack([np.array(rep_xs_all), np.array(rep_ys_all)])

cluster_candidates = {}

for cluster_id in range(k):
    cluster_mask = labels == cluster_id
    cluster_points = X[cluster_mask]
    global_indices = np.where(cluster_mask)[0]   # index directly into cell_coords

    dists = cdist(cluster_points, kmeans.cluster_centers_[cluster_id].reshape(1, -1)).flatten()
    order = np.argsort(dists)
    top_n_global = global_indices[order[:N_CANDIDATES]]

    cluster_candidates[cluster_id] = [int(i) for i in top_n_global]

# --------------------------------------------------
# 8. Greedy / shuffle-greedy selection with min distance constraint
# --------------------------------------------------
min_dist = 2000   # metres — adjust to your required gauge spacing
existing_coords = np.empty((0, 2))   # or np.array of shape (m, 2) with real gauge coords

solution = shuffle_greedy(cluster_candidates, cell_coords, existing_coords,
                           min_dist, n_iterations=500)

_, failed = greedy_selection(cluster_candidates, cell_coords, existing_coords, min_dist)
if failed:
    print("Clusters with no feasible candidate:", failed)

# --------------------------------------------------
# 9. Map chosen indices back to row/col + attributes
# --------------------------------------------------
selected_records = []
for cluster_id, idx in solution.items():
    row, col = valid_rows[idx], valid_cols[idx]
    x, y = cell_coords[idx]
    selected_records.append({
        "cluster": cluster_id,
        "row": row, "col": col,
        "x": x, "y": y,
        "dem": dem_arr[row, col],
        "slope": slope_arr[row, col],
        "aspect": aspect_arr[row, col],
    })

selected_df = pd.DataFrame(selected_records)
selected_df

# --------------------------------------------------
# 10. Plot final spacing-constrained representatives
# --------------------------------------------------
fig, ax = plt.subplots(figsize=(6, 5))
label_xr.plot(ax=ax, cmap=cmap, norm=norm, add_colorbar=True,
              cbar_kwargs={"ticks": np.arange(k), "label": "cluster"})
catchment_Dolwen.boundary.plot(ax=ax, color="black")
ax.scatter(selected_df["x"], selected_df["y"], c="black", edgecolor="white",
           s=80, marker="*", zorder=5)
ax.axis('off')
ax.set_title(f'K-means clusters (k={k}) — spacing-constrained representatives')
```
